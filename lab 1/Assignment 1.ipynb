```python
import cv2

def process_image(input_path, output_path):
    
    img = cv2.imread(input_path)
    if img is None:
        print(f"Error: Could not read the image at '{input_path}'. Check the file path.")
        return

    print("Image read successfully!")

    cv2.imshow('Image Display Window', img)
    print("Press any key in the image window to continue...")
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()

    success = cv2.imwrite(output_path, img)
    
    if success:
        print(f"Image successfully saved as '{output_path}'.")
    else:
        print("Error: Could not save the image.")

if __name__ == "__main__":
    process_image('input_image.jpeg', 'output_image.png')
    



    
```

```python
import cv2

def convert_image_formats(input_path):
    img = cv2.imread(input_path)
    
    if img is not None:
        cv2.imwrite('output_image.png', img)
        cv2.imwrite('output_image.tiff', img)
        cv2.imwrite('output_image.bmp', img)

if __name__ == "__main__":
    convert_image_formats('input_image.jpeg')
```

```python
import cv2

def resize_image(input_path, output_path, width, height):
    img = cv2.imread(input_path)
    
    if img is not None:
        resized_img = cv2.resize(img, (width, height))
        cv2.imwrite(output_path, resized_img)

if __name__ == "__main__":
    resize_image('input_image.jpg', 'resized_image.jpg', 800, 600)
    
```

```python
import cv2

def convert_to_grayscale(input_path, output_path):
    img = cv2.imread(input_path)
    
    if img is not None:
        gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        cv2.imwrite(output_path, gray_img)

if __name__ == "__main__":
    convert_to_grayscale('input_image.jpeg', 'grayscale_image.jpg')
```

```python
import cv2

def play_video(video_path):
    cap = cv2.VideoCapture(video_path)

    while cap.isOpened():
        ret, frame = cap.read()
        
        if not ret:
            break  
            
        cv2.imshow('Video Playback', frame)
        
        if cv2.waitKey(25) & 0xFF == ord('q'):
            break
    
    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    play_video('input_video.mp4')
```

```python
import cv2
import os

def extract_frames(video_path, output_folder):
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    cap = cv2.VideoCapture(video_path)
    frame_count = 0

    while cap.isOpened():
        ret, frame = cap.read()
        
        if not ret:
            break
            
        frame_filename = os.path.join(output_folder, f"frame_{frame_count:04d}.jpg")
        cv2.imwrite(frame_filename, frame)
        frame_count += 1

    cap.release()
                
if __name__ == "__main__":
    extract_frames('input_video.mp4', 'extracted_frames') 
```
